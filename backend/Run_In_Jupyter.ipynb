### 3️⃣ Server start karo
Server background me chalega, notebook block nahi hoga.

```python
import uvicorn, threading, time

config = uvicorn.Config("app.main:app", host="127.0.0.1", port=8000, log_level="warning")
server = uvicorn.Server(config)
threading.Thread(target=server.run, daemon=True).start()
time.sleep(2)

print("🚀 Server chal raha hai! Browser me kholo:")
print("   http://localhost:8000/auth.html   (Login / Register)")
print("   http://localhost:8000/            (Landing page)")
print("   http://localhost:8000/docs        (API docs)")
```

### 4️⃣ (Optional) Yahin se test karo
Neeche wali cell server ko check karti hai aur ek naya account bana ke dikhati hai.

```python
import urllib.request, json

def call(path, body=None, token=None):
    url = "http://localhost:8000" + path
    data = json.dumps(body).encode() if body else None
    req = urllib.request.Request(url, data=data, method="POST" if body else "GET")
    req.add_header("Content-Type", "application/json")
    if token:
        req.add_header("Authorization", "Bearer " + token)
    try:
        with urllib.request.urlopen(req) as r:
            return r.status, json.load(r)
    except urllib.error.HTTPError as e:
        return e.code, json.load(e)

print("health:", call("/health"))
print("login demo:", call("/auth/login", {"email": "demo@aitradingmaster.com", "password": "Demo@12345"})[0])
```
